```python
####Dynamic Prompting
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate 
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")#loading ai model
out=StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"Your a translater and translate input into {language} "},{"role":"user","content":"{query}"}
])
```

```python

final_prompts=prompts.invoke({"language":"hinglish","query":"I love you"})
res=llm.invoke(final_prompts)
print(res.content)
```

```python
res=llm.invoke(prompts.invoke({"language":"hinglish","query":"I love you"}))
print(res)
```

```python
####Chains
```

```python
chains=prompts|llm|out
res=chains.invoke({"language":"hinglish","query":"I love you"})
res
```
