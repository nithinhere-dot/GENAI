```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    name:str
    age:int
```

```python
user=User(name="Nithin",age=21)
user
```

Field Validation

```python
from pydantic import Field
```

```python
class Product(BaseModel):
    title:str=Field(max_length=10)
    des:str=Field(min_length=2)
    price:int=Field(gt=0)
```

```python
prod=Product(title="Milk",des="Drinking",price=20)
```

```python
prod
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    prodcut:Product
    
```

```python
order=Order(id="adjnj1",prodcut=prod)
```

```python
order
```

```python
from typing import List
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:List[Product]
    comment:str=Field(default="No Comment")
```

```python
cart=Carts(id="idsji1",items=[prod])
```

```python
cart
```

```python
from typing import Annotated
##for common validator
```

```python
commonValidator =Annotated[List[Product],"This is common Validator"]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:commonValidator
    comment:str=Field(default="No Comment")
```

```python
class Order(BaseModel):
    products:commonValidator
```
