```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
        it will return the sum of two numbers

        args:
            a:number One
            b:number Two
    """
    return a+b
```

```python
add_numbers.invoke({"a":12,"b":10})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent #connect tools and llm model
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers], #must be tools
    system_prompt="Your a math teacher always use tools for calculation"
)
```

```python
res=agent.invoke({"messages":[{"role":"user","content":"What is 2+3"}]})
print(res["messages"][-1].content)
```
